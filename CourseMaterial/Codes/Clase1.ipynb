### Cargar una imagen

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

```python
imagen = cv2.imread('/content/img_resize_460.png')
imagen_rgb = cv2.cvtColor(imagen, cv2.COLOR_BGR2RGB)
plt.imshow(imagen_rgb)
plt.show()
```

### Canales de la imagen

```python
# Separar la imagen RGB en sus tres canales individuales: rojo (r), verde (g) y azul (b)
r, g, b = cv2.split(imagen_rgb)

# Crear una matriz de ceros del mismo tamaño que un canal
# Esto nos servirá para "apagar" los otros canales al crear imágenes de un solo color
cero = np.zeros_like(r)

# Crear una imagen donde solo el canal rojo está activo
# Combinamos el canal rojo con ceros en verde y azul
rojo = cv2.merge([r, cero, cero])

# Crear una imagen donde solo el canal verde está activo
verde = cv2.merge([cero, g, cero])

# Crear una imagen donde solo el canal azul está activo
azul = cv2.merge([cero, cero, b])
```

Ahora tenemos:
- **rojo**: solo el componente rojo de la imagen
- **verde**: solo el componente verde de la imagen
- **azul**: solo el componente azul de la imagen

### Gráfico de los canales

```python
fig, axs = plt.subplots(1, 4, figsize=(20, 5))

axs[0].imshow(imagen_rgb)
axs[0].set_title('Imagen Original')

axs[1].imshow(rojo)
axs[1].set_title('Componente Rojo')

axs[2].imshow(verde)
axs[2].set_title('Componente Verde')

axs[3].imshow(azul)
axs[3].set_title('Componente Azul')

plt.show()
```
